#CartPole-v1
import gymnasium as gym

# Tạo môi trường
env = gym.make("CartPole-v1", render_mode="human")

# Đặt lại môi trường
obs, info = env.reset()

for _ in range(500):
    action = env.action_space.sample() # hành động ngẫu nhiên
    obs, reward, done, truncated, info = env.step(action)
    if done or truncated:
        obs, info = env.reset()

env.close()

1.	Môi trường
- Với RL thì bước đầu tiên luôn là tạo môi trường cho toàn hệ thống, việc tạo môi trường sẽ giúp cho hệ thống trở nên tốt hơn. Và bằng cách thay đổi các môi trường khác nhau hệ thống hoàn thiện hơn vì mỗi môi trường sẽ có quy luật và cách xử lý khác nhau.
2.	Khởi tạo episode
- obs, info = env.reset(): là bắt đầu 1 lượt mới với 2 trạng thái ban đầu là observation và information. 
- observation chỉ quan sát ban đầu của agent khi khởi tạo 1 episode mới
- information là thông tin ban đầu môi trường cung cấp có agent
3.	Thực hiện action
- action = env.action_space.sample(): thực hiện 1 hành động ngẫu nhiên trong action_space
- obs, reward, done, truncated, info = env.step(action): lựa chọn những action để agent thực hiện. Sau đó xét done or truncated (có thể là terminated) để reset lại từ đầu
+ Truncated: là viêc kết thúc do vi phạm luật chơi
+ Terminated: là kết thúc do thua như kiểu là gậy đổ, đâm vào vật cản hoặc là rơi xuống hố
+ Done: là kết thúc vì bất kì lí do gì, dù là quá thời gian, mất quá số lượt quy định hay rơi xuống hố
- env.close(): đóng quá trình hiển thị, đóng môi trường, nhưng chương trình vẫn tiếp tục

II.	Thử môi trường khác
- Đối với môi trường “MountainCar-v0” độ khó của nó lớn hơn “CartPole-v1”, bởi vì xe phải lấy đà qua lại để tới được đỉnh, nên nó có nhiều trạng thái hơn, reward cũng hiếm hơn
- print(env.action_space): số hành động mà agent có thể thực hiện
- print(env.observation_space): thông tin mà agent nhận được từ môi trường, và đối với “MountainCar-v0” thì sẽ có 2 môi trường là position và velocity
